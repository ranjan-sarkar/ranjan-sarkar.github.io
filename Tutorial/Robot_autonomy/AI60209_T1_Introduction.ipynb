# 1. **NumPy (Numerical Python)**

**NumPy** is the core library for scientific computing in Python. It provides high-performance multi-dimensional array objects (`ndarray`) and vectorized mathematical operations essential for numerical computation, matrix operations, and state representation in Reinforcement Learning.

### 1.1 Array Creation from Python Lists
You can convert standard Python lists into 1D vectors or 2D matrices.

````python
import numpy as np

# Create a 1D array (Vector)
arr1 = np.array([1, 2, 3, 4, 5])
print("1D Array:", arr1)
print("Type:", type(arr1))
print("Data Type (dtype):", arr1.dtype)
print("Shape:", arr1.shape)
print("Dimensions (ndim):", arr1.ndim)
````

### 1.2 Multi-dimensional Arrays (Matrices)
Matrices are 2D or N-D arrays.

````python
# Create a 2D array (Matrix)
arr2 = np.array([[1, 2, 3], [4, 5, 6]])
print("2D Array:\n", arr2)
print("Shape of arr2:", arr2.shape)
print("Dimensions (ndim):", arr2.ndim)
````

### 1.3 Built-in Creation Functions
NumPy provides functions like `zeros()`, `ones()`, and `eye()` to generate pre-filled matrices.

````python
# Array of zeros
zeros_arr = np.zeros((2, 3))
print("Zeros Array (2x3):\n", zeros_arr)

# Array of ones
ones_arr = np.ones((3, 2))
print("\nOnes Array (3x2):\n", ones_arr)

# Identity matrix
identity_mat = np.eye(3)
print("\nIdentity Matrix (3x3):\n", identity_mat)
````

### 1.4 Generating Numerical Ranges
Use `arange()` for step-based sequences and `linspace()` for evenly spaced numbers over an interval.

````python
# Step-based range: start=0, stop=10 (exclusive), step=2
range_arr = np.arange(0, 10, 2)
print("Array from arange:", range_arr)

# Linearly spaced array: start=0, stop=1 (inclusive), num=5
linspace_arr = np.linspace(0, 1, 5)
print("Array from linspace:", linspace_arr)
````

### 1.5 Basic Element-wise Operations
NumPy operations are vectorized, performing calculations element-by-element without explicit Python `for` loops.

````python
arr_a = np.array([1, 2, 3])
arr_b = np.array([4, 5, 6])

print("Vector A:", arr_a)
print("Vector B:", arr_b)
print("Element-wise Addition (A + B):", arr_a + arr_b)
print("Element-wise Multiplication (A * B):", arr_a * arr_b)
print("Power (A ** 2):", arr_a ** 2)
print("Scalar Multiplication (A * 3):", arr_a * 3)
````

### 1.6 Vector & Matrix Multiplication
Vector dot products can be computed with `np.dot()`, while matrix multiplication uses the `@` operator or `np.dot()`.

````python
# Vector Dot Product
dot_product = np.dot(arr_a, arr_b)
print("Dot Product (A . B):", dot_product)

# Matrix Multiplication
mat1 = np.array([[1, 2], [2, 1]])
mat2 = np.array([[2, 0], [1, 2]])

matrix_mult = mat1 @ mat2
print("\nMatrix 1:\n", mat1)
print("Matrix 2:\n", mat2)
print("Matrix Multiplication (mat1 @ mat2):\n", matrix_mult)

# Matrix Transpose
print("\nTransposed Matrix 1 (mat1.T):\n", mat1.T)
````

### 1.7 Linear Algebra: Eigenvalues & Eigenvectors
The `numpy.linalg` module provides linear algebra routines such as spectral decomposition.

````python
import numpy.linalg as la

# Spectral Decomposition (Eigenvalues & Eigenvectors)
eigvalues, eigenvectors = la.eig(mat1)
print("Eigenvalues:", eigvalues)
print("\nEigenvectors:\n", eigenvectors)
````

### 1.8 Linear Algebra: Singular Value Decomposition (SVD)
Decomposing a matrix into left singular vectors $U$, singular values $S$, and right singular vectors $V^T$.

````python
# Singular Value Decomposition
U, S, VT = la.svd(mat1)
print("Left Singular Vectors (U):\n", U)
print("Singular Values (S):", S)
print("Right Singular Vectors Transposed (V^T):\n", VT)
````

### 1.9 Indexing & Slicing
Accessing elements, specific rows, columns, or sub-matrices.

````python
arr = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])
print("Original Matrix:\n", arr)

print("\nElement at (row 0, col 1):", arr[0, 1])
print("First row:", arr[0, :])
print("Second column:", arr[:, 1])
print("Sub-matrix (rows 0-1, cols 1-2):\n", arr[0:2, 1:])
````

### 1.10 Boolean Indexing & Masking
Filtering array elements based on logical conditions.

````python
# Create a boolean mask
mask = arr > 50
print("Boolean Mask (arr > 50):\n", mask)

# Extract elements satisfying condition
print("Elements > 50:", arr[mask])
````

### 1.11 Reshaping & Flattening
Modifying array dimensions without altering underlying data.

````python
arr1d = np.arange(1, 10)
print("Original 1D Array:", arr1d)

# Reshape into a 3x3 matrix
reshaped_3x3 = arr1d.reshape(3, 3)
print("\nReshaped 3x3 Matrix:\n", reshaped_3x3)

# Flatten back to 1D
flattened = reshaped_3x3.flatten()
print("\nFlattened Array:", flattened)

# Infer dimensions using -1
inferred = arr1d.reshape(3, -1)
print("\nInferred Reshape (3 rows):\n", inferred)
````

### 1.12 Universal Functions (ufuncs) & Aggregations
Fast element-wise operations and summary statistics.

````python
data = np.array([1, 4, 9, 16, 25])
print("Original Data:", data)

print("Square Root:", np.sqrt(data))
print("Exponential:", np.exp(data))
print("Sine:", np.sin(data))

print("\n--- Summary Statistics ---")
print("Sum:", np.sum(data))
print("Mean:", np.mean(data))
print("Standard Deviation:", np.std(data))
print("Min:", np.min(data))
print("Max:", np.max(data))
````

Used to display the distribution of a dataset. It groups values into bins and counts how many values fall into each bin.

````python
np.random.seed(42)
data_hist = np.random.randn(1000)

plt.figure(figsize=(8, 4.5))
n, bins, patches = plt.hist(
                         data_hist, bins=30,
                         color='#6C5CE7', edgecolor='black',
                         alpha=0.75, density=True
                    )

# Theoretical Normal PDF overlay
mu, std = 0, 1
p = (1 / (std * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((bins - mu) / std) ** 2)
plt.plot(bins, p, color='#D63031', linewidth=2.5, label='Theoretical PDF')

plt.title('Histogram of Normally Distributed Data', fontsize=14, fontweight='bold')
plt.xlabel('Random Variable $X$', fontsize=12)
plt.ylabel('Probability Density', fontsize=12)
plt.grid(axis='y', linestyle=':', alpha=0.6)
plt.legend(fontsize=11)
plt.tight_layout()
plt.show()
````

# 3. **SymPy (Symbolic Python)**

**SymPy** is a Python library for **symbolic mathematics**. It performs exact calculus, algebra, matrix derivatives, and equation solving symbolically, useful for deriving analytical gradients and bellman updates.

### 3.1 Defining Symbolic Variables & Vector Matrices

````python
from sympy import symbols, Matrix, Eq, solve, diff, hessian, integrate, plot
try:
    from IPython.display import display
except ImportError:
    pass

# Define symbolic variables x and y
x, y = symbols('x y')
var = Matrix([x, y])

print("Symbolic Vector Variable:")
display(var)
````

### 3.2 Defining Multi-variable Expressions
Consider the objective function $f(x, y) = x^2 + y^3$:

````python
exp = x**2 + y**3
print("Symbolic Expression f(x, y):")
display(exp)
````

### 3.3 Symbolic Differentiation (Gradient & Hessian)
Compute the analytical Gradient vector $\nabla f$ and Hessian matrix $H(f)$.

````python
# Gradient vector (first derivatives)
grad = diff(exp, var)
print("Gradient Vector:")
display(grad)

# Hessian matrix (second derivatives)
hess = hessian(exp, var)
print("Hessian Matrix:")
display(hess)
````

### 3.4 Evaluating Gradient at a Numerical Point
Substitute numerical values $(x=1, y=1)$ into the symbolic gradient.

````python
# Substitute (x=1, y=1)
grad_evaluated = grad.subs({x: 1, y: 1})
print("Gradient evaluated at (x=1, y=1):")
display(grad_evaluated)
````

### 3.5 Symbolic Integration
Compute indefinite integral $\int (x^2 + y^3) \, dx$:

````python
int_exp = integrate(exp, x)
print("Integral with respect to x:")
display(int_exp)
````

### 3.6 Solving Algebraic Equations
Solve the quadratic equation $x^2 - 5x + 6 = 0$:

````python
# Define equation Eq(lhs, rhs)
eq = Eq(x**2 - 5*x + 6, 0)
print("Equation:")
display(eq)

# Solve for x
solutions = solve(eq, x)
print("Analytical Solutions for x:", solutions)
````

### 3.7 Symbolic Plotting
Directly plot a symbolic function using SymPy's built-in plotting capability.

````python
plot(x**2 - 5*x + 6, (x, -1, 5), title='$f(x) = x^2 - 5x + 6$', show=True);
````

# 4. **CasADi (Nonlinear Optimization)**

`CasADi (Computer Algebra System and Automatic Differentiation)` is a tool for numerical optimization and algorithmic differentiation, widely used in Optimal Control, Model Predictive Control (MPC), and Trajectory Optimization.

### 4.1 Optimization Problem Formulation
We aim to solve the following Constrained Nonlinear Programming (NLP) problem:

$$\min_{x,y} \quad (y - x^2)^2$$

$$\begin{align*}
\text{subject to:} \quad & x^2 + y^2 = 1 \\
& x + y \ge 1
\end{align*}$$

### 4.2 Initializing the Opti Stack & Decision Variables
The `Opti` class in CasADi provides a high-level syntax to define variables, objectives, and constraints.

* `opti.variable()`: Creates a decision variable.

* `opti.minimize()`: Defines the objective.

* `opti.subject_to()`: Adds constraints to the problem

````python
import casadi as cd

# Initialize the Opti stack
opti = cd.Opti()

# Declare Decision Variables
x = opti.variable()
y = opti.variable()

# Define Objective Function
obj = (y - x**2)**2
opti.minimize(obj)

# Define Constraints
opti.subject_to(x**2 + y**2 == 1)
opti.subject_to(x + y >= 1)

print("Opti stack setup completed successfully.")
````

### 4.3 Solver Configuration & Execution
We configure `IPOPT (Interior Point OPTimizer)`, a standard solver for large-scale nonlinear optimization problems.

`opti.solve()`: This command triggers the solver. It will compute the gradients and Hessians automatically and iterate until it finds a local minimum.

````python
# Select IPOPT solver (quiet mode)
opti.solver('ipopt', {'ipopt.print_level': 0, 'print_time': 0})

try:
    sol = opti.solve()
    print("Solver converged successfully!")
except Exception as e:
    print("Solver failed to converge:", e)
````

### 4.4 Solution Extraction
Extract numerical values of optimal variables $x^*$, $y^*$ and optimal cost.

The `solve()` function returns a solution object. To get the actual numerical value of a variable, you must use `sol.value(variable_name)`.

````python
# Extract values from solution object
x_opt = sol.value(x)
y_opt = sol.value(y)
cost_opt = sol.value(obj)

print(f"Optimal x*        : {x_opt:.4f}")
print(f"Optimal y*        : {y_opt:.4f}")
print(f"Optimal Objective : {cost_opt:.6f}")
````

## 5.1 Environment Concept

### Environment
We say that the agent interacts with the environment. What is the meaning of this?

In computer science, agent refers to your computational unit and environment refers to everythings else in a Cyber Physical systems.

For the perspective of any other engineering branch, you can consider that the environment consist of your physical system and its connection with the agent. The agent can be considered a decision making device, which is a controller. 

In Open AI Gym, we have several environments such as
1. Cart-pole
2. Mountain car

An RL agent interacts with an environment in a closed-loop fashion:

1. **State / Observation ($s_t$)**: What the agent observes about the environment.
2. **Action ($a_t$)**: The command sent by the agent/controller.
3. **Reward ($r_t$)**: Scalar feedback signal indicating how good the action was.
4. **Next State ($s_{t+1}$)**: The resulting environment state.
5. **Termination / Truncation**: Signals indicating whether the episode ended.

## 5.2 CartPole-v1 Environment
In **CartPole-v1**, a pole is attached by an un-actuated joint to a cart moving along a frictionless track. The goal is to balance the pole upright by pushing the cart left or right.

````python
import gymnasium as gym
import numpy as np
import time

# Initialize environment
env = gym.make('CartPole-v1', render_mode='human')
print("Environment Created: CartPole-v1")
````

````python
# Reset environment to start a new episode
env.reset()
````

````python
# Close the pygame window
env.close()
````

### 5.3 Observation Space
The observation space is a `Box(4,)` representing 4 continuous state variables:
1. **Cart Position**: $[-4.8, 4.8]$
2. **Cart Velocity**: $(-\infty, \infty)$
3. **Pole Angle**: [$-24^\circ, 24^\circ$] ($\approx [-0.418 \text{ rad}, 0.418 \text{ rad}]$)
4. **Pole Angular Velocity**: $(-\infty, \infty)$

````python
env = gym.make('CartPole-v1')
obs_space = env.observation_space

print("Observation Space Shape:", obs_space.shape)
print("Observation High Bounds:", obs_space.high)
print("Observation Low Bounds:", obs_space.low)

env.close()
````

### 5.4 Action Space
In CartPole, the action space is `Discrete(2)`:
* `0`: Push Cart Left
* `1`: Push Cart Right

````python
env = gym.make('CartPole-v1')
action_space = env.action_space

print("Action Space Type:", type(action_space))
print("Total number of actions (n):", action_space.n)

# Sample random valid actions
sample_action = action_space.sample()
print("Sampled Random Action:", sample_action)

env.close()
````

#### What else agent gets after each step

Every time the agent takes a step, the environment returns five values:

1. `Observation`: The new state of the world.

2. `Reward`: Usually +1 for each time the pole stays up.

3. `Terminated`: True if the pole falls or the cart moves too far.

4. `Truncated`: True if the episode timed out (e.g., reached 500 steps).

5. `Info`: Extra debugging data.

### 5.5 Baseline Agent 1: Fixed Policy (Always Push Left)
Let's test an agent that always chooses `action = 0` (Push Left).

````python
import time

env = gym.make('CartPole-v1', render_mode="human")
env.reset()
action = 0      # ACTION: moving left

for i in range(50):    

    obs, reward, done, *_ = env.step(action)
    print("step", i, obs, reward, done, action)
    
    time.sleep(0.1)

env.close()
````

### 5.6 Baseline Agent 2: Fixed Policy (Always Push Right)
Let's test an agent that always chooses `action = 1` (Push Right).

````python
env = gym.make('CartPole-v1', render_mode="human")
env.reset()
action = 1      # ACTION: moving right      

# Game configs
total_reward = 0.0
total_steps = 0

done = False

for i in range(100):    

    obs, reward, terminated, truncated, info = env.step(action)

    total_reward += reward
    total_steps += 1
    time.sleep(0.1)

    # If the episode is up, then start another one
    if terminated:
        env.reset()
    print(f"Episode completed in {total_steps} steps.\nTotal rewards accumulated: {total_reward}")    

env.close()
````

### 5.7 Baseline Agent 3: Uniform Random Policy
Choosing random actions `env.action_space.sample()` at each timestep.

````python
env = gym.make('CartPole-v1', render_mode="human")
env.reset()

for i in range(500):

    # env.action_space.sample() produces either 0 (left) or 1 (right).
    action = env.action_space.sample()
    observation, reward, terminated, truncated, info = env.step(action)
    time.sleep(0.05)
    print("step", i, observation, reward, terminated, truncated, action)

    # If the episode is up, then start another one
    if terminated:
        env.reset()

env.close()
````

### 5.10 Transition Test on MountainCar
Reset the environment and execute 1 step with a random action, and print next observation.

````python
# reset the environment and see the initial observation
env = gym.make('MountainCar-v0')

obs, info = env.reset()
print("The initial observation is {}".format(obs))

# Sample a random action from the entire action space
random_action = env.action_space.sample()

# # Take the action and get the new observation space
new_obs, reward, terminated, truncated, info = env.step(random_action)
print("The new observation is {}".format(new_obs))
env.close()
````

### 5.11 Random Agent Simulation Loop on MountainCar

````python
import time 

# Number of steps you run the agent for 
num_steps = 500

env = gym.make('MountainCar-v0', render_mode="human")
obs = env.reset()

for step in range(num_steps):
    
    action = env.action_space.sample()
    
    # apply the action
    obs, reward, terminated, truncated, info = env.step(action)
    
    # Render the env
    env.render()

    # Wait a bit before the next frame unless you want to see a crazy fast video
    time.sleep(0.001)
    
    # If the episode is up, then start another one
    if terminated:
        env.reset()

# Close the env
env.close()
````
