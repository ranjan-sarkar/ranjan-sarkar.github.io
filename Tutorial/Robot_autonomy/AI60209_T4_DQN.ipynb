  <h2 style="margin: 0; font-size: 1.35rem; font-weight: 700; letter-spacing: -0.01em;">
    1. Motivation: From Tabular Q-Learning to Deep Q-Networks
  </h2>
</div>

In **Tutorial 3**, we solved `MountainCar-v0` by discretizing continuous position and velocity into a $20 \times 20 \times 3$ lookup table:

$$Q(s_t, a_t) \leftarrow Q(s_t, a_t) + \alpha \left[ R_{t+1} + \gamma \max_{a'} Q(s_{t+1}, a') - Q(s_t, a_t) \right]$$

### Why Move from a Table to a Neural Network?
1. **No Manual Discretization:** Instead of hand-crafting arbitrary bin boundaries, a neural network $Q(s, a; \theta) \approx Q^*(s, a)$ accepts continuous states directly (such as cart position, cart velocity, pole angle, and angular velocity).
2. **Generalization:** Learning in one state smoothly updates network weights, improving value estimates for nearby unvisited states.
3. **Scalability:** Tabular methods explode exponentially in higher-dimensional robotics, while neural networks scale to complex continuous spaces.

### DeepMind's Two Core Stabilizations (2015)
- **Experience Replay:** Stores transitions $(s, a, r, s', d)$ in a buffer and samples random mini-batches $\to$ breaks temporal correlation and restores i.i.d. training.
- **Target Network ($Q_{\text{target}}$):** A slowly updated network copy used to compute Bellman bootstrap targets $\to$ eliminates the "moving target" problem.

<div style="background: linear-gradient(90deg, rgba(56, 189, 248, 0.18) 0%, rgba(99, 102, 241, 0.10) 55%, transparent 100%); border-left: 5px solid #38bdf8; padding: 10px 18px; border-radius: 8px; margin: 24px 0 14px 0;">
  <h2 style="margin: 0; font-size: 1.35rem; font-weight: 700; letter-spacing: -0.01em;">
    2. Environment Setup: `CartPole-v1`
  </h2>
</div>

A pole is attached by an un-actuated joint to a cart moving along a frictionless track. The system is controlled by applying a force of $+1$ or $-1$ to the cart. The pendulum starts upright, and the goal is to prevent it from falling over:
- **State ($s \in \mathbb{R}^4$):** Continuous vector `[cart_position, cart_velocity, pole_angle, pole_angular_velocity]`
  - Cart Position ($x$): $[-4.8, 4.8]$ (termination limit: $\pm 2.4$)
  - Cart Velocity ($\dot{x}$): $[-\infty, \infty]$
  - Pole Angle ($\theta$): $[-0.418 \text{ rad}, 0.418 \text{ rad}]$ (termination limit: $\pm 12^\circ \approx \pm 0.2095 \text{ rad}$)
  - Pole Angular Velocity ($\dot{\theta}$): $[-\infty, \infty]$
- **Action ($a \in \{0, 1\}$):** `0`: Push Cart Left, `1`: Push Cart Right
- **Reward & Termination:** The agent receives a reward of $+1$ for every step taken where the pole remains balanced. The episode terminates if $|\theta| > 12^\circ$ or $|x| > 2.4$. The episode is truncated (solved) upon reaching a threshold of 500 steps in `CartPole-v1`.

```python
import random
from collections import deque, namedtuple
from itertools import count

import gymnasium as gym
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim

# Configure Matplotlib styling: serif fonts, high-DPI rendering, and publication aesthetics
plt.rcParams['font.family'] = 'serif'
plt.rcParams['font.serif'] = ['Times New Roman', 'DejaVu Serif', 'serif']
plt.rcParams['mathtext.fontset'] = 'stix'
plt.rcParams['figure.dpi'] = 150
plt.rcParams['savefig.dpi'] = 150
plt.rcParams['axes.linewidth'] = 1.2

# Set up matplotlib for high quality and live interactive plotting
is_ipython = 'inline' in matplotlib.get_backend()
if is_ipython:
    from IPython import display
    from IPython import get_ipython
    ipython = get_ipython()
    if ipython is not None:
        ipython.run_line_magic('config', "InlineBackend.figure_format = 'retina'")

# Enable interactive mode for real-time plotting
plt.ion()
```

```python
# Create CartPole environment with simulation window open
env = gym.make("CartPole-v1", render_mode="human")

env.reset()

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device} | Gymnasium version: {gym.__version__}")
print(f"Observation space: {env.observation_space}")
print(f"Action space: {env.action_space}")
```

  <h2 style="margin: 0; font-size: 1.35rem; font-weight: 700; letter-spacing: -0.01em;">
    3. Experience Replay Buffer
  </h2>
</div>

Transitions are stored as $e_t = (s_t, a_t, r_{t+1}, s_{t+1}, \text{done})$.

Uniform random sampling from a circular buffer:
1. **Breaks temporal correlation** between consecutive environment steps.
2. **Reuses transitions** efficiently across multiple gradient updates.


```python
# Define transition namedtuple
Transition = namedtuple('Transition',
                        ('state', 'action', 'next_state', 'reward', 'done'))

class ReplayMemory:
    """Fixed-capacity circular buffer to store and sample experience transitions."""

    def __init__(self, capacity: int = 20_000):
        self.memory = deque([], maxlen=capacity)    # deque: double-ended queue

    def push(self, *args):
        """Save a transition: (state, action, next_state, reward, done)"""
        self.memory.append(Transition(*args))

    def sample(self, batch_size: int):
        """Sample a random mini-batch of transitions."""
        return random.sample(self.memory, batch_size)

    def __len__(self):
        return len(self.memory)
```

<div style="background: linear-gradient(90deg, rgba(56, 189, 248, 0.18) 0%, rgba(99, 102, 241, 0.10) 55%, transparent 100%); border-left: 5px solid #38bdf8; padding: 10px 18px; border-radius: 8px; margin: 24px 0 14px 0;">
  <h2 style="margin: 0; font-size: 1.35rem; font-weight: 700; letter-spacing: -0.01em;">
    4. Deep Q-Network Architecture
  </h2>
</div>

The network maps continuous state $s = [x, \dot{x}, \theta, \dot{\theta}] \in \mathbb{R}^4$ directly to Q-values for all 2 discrete actions in a single forward pass:

$$\text{State } s \in \mathbb{R}^4 \longrightarrow \text{Linear}(128) \to \text{ReLU} \longrightarrow \text{Linear}(128) \to \text{ReLU} \longrightarrow [Q(s, \text{Left}), Q(s, \text{Right})]$$

```python
class DQN(nn.Module):
    """Deep Q-Network mapping state observations to discrete action Q-values."""

    # Q-network - a simple feedforward neural network with 2 hidden layers
    def __init__(self, n_observations = 4, n_actions = 2):
        super(DQN, self).__init__()
        self.layer1 = nn.Linear(n_observations, 128)
        self.layer2 = nn.Linear(128, 128)
        self.layer3 = nn.Linear(128, n_actions)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        x = F.relu(self.layer1(x))
        x = F.relu(self.layer2(x))
        return self.layer3(x)
```

<div style="background: linear-gradient(90deg, rgba(56, 189, 248, 0.18) 0%, rgba(99, 102, 241, 0.10) 55%, transparent 100%); border-left: 5px solid #38bdf8; padding: 10px 18px; border-radius: 8px; margin: 24px 0 14px 0;">
  <h2 style="margin: 0; font-size: 1.35rem; font-weight: 700; letter-spacing: -0.01em;">
    5. Hyperparameters & Exploration
  </h2>
</div>

- **$\epsilon$-Greedy Policy:** Balances exploration (random action with probability $\epsilon$) and exploitation (greedy action $\operatorname{argmax}_a Q$ with probability $1 - \epsilon$). $\epsilon$ decays from $1.0$ down to $0.05$.
- **Soft Target Update (Polyak Averaging):** Slowly tracks the online network at every step:
  $$\theta_{\text{target}} \leftarrow \tau \theta_{\text{online}} + (1 - \tau) \theta_{\text{target}} \quad (\tau = 0.005)$$


```python
# Hyperparameters
BATCH_SIZE = 128         # Mini-batch size sampled from replay memory
GAMMA = 0.99            # Discount factor for future rewards
EPS_START = 0.9        # Starting exploration rate
EPS_END = 0.05          # Minimum exploration rate
EPS_DECAY = 0.98        # Decay factor applied per episode
TAU = 0.005             # Soft update rate (Polyak averaging)
LR = 1e-3               # AdamW learning rate

# Recreate a non-rendering environment so this cell is safe after env.close()
env = gym.make("CartPole-v1")

# Initialize Environment & Models
n_actions = env.action_space.n
initial_state, _ = env.reset()
n_observations = len(initial_state)

policy_net = DQN(n_observations, n_actions).to(device)
target_net = DQN(n_observations, n_actions).to(device)
target_net.load_state_dict(policy_net.state_dict())         # initialize target network with policy network weights
target_net.eval()

optimizer = optim.AdamW(policy_net.parameters(), lr=LR, amsgrad=True)
criterion = nn.SmoothL1Loss()  # Huber loss
memory = ReplayMemory(20_000)

current_epsilon = EPS_START


def select_action(state: torch.Tensor, epsilon: float = None) -> torch.Tensor:
    """Select an action using epsilon-greedy policy."""
    eps = current_epsilon if epsilon is None else epsilon
    if random.random() > eps:
        with torch.no_grad():
            return policy_net(state).argmax(dim=1).view(1, 1)
    else:
        action = env.action_space.sample()
        return torch.tensor([[action]], device=device, dtype=torch.long)


episode_durations = []
fig, ax = None, None
line_raw, line_avg = None, None
display_handle = None


def plot_durations(show_result=False):
    """Plot live episode duration/reward and smooth 50-episode moving average without flicker."""
    global fig, ax, line_raw, line_avg, display_handle
    durations_t = torch.tensor(episode_durations, dtype=torch.float)
    episodes = np.arange(1, len(durations_t) + 1)

    # Initialize figure & axes once on first call
    if fig is None or ax is None:
        plt.ioff()
        fig, ax = plt.subplots(figsize=(9, 4.8), dpi=150)
        ax.set_xlim(0, max(200, len(episodes)))
        ax.set_ylim(0, 300)  
        ax.set_xlabel('Episode', fontsize=12, fontweight='bold')
        ax.set_ylabel('Duration / Reward', fontsize=12, fontweight='bold')
        # ax.axhline(y=500, color='#16a34a', linestyle='--', linewidth=1.5, label='Max Duration (500)', alpha=0.8)
        line_raw, = ax.plot([], [], color='#38bdf8', alpha=0.35, linewidth=1.2, label='Episode Duration')
        line_avg, = ax.plot([], [], color='#0284c7', linewidth=2.5, label='50-Ep Moving Avg')
        ax.legend(loc='upper left', framealpha=0.9, fontsize=10)
        ax.grid(True, linestyle='--', alpha=0.5, linewidth=0.8)
        fig.tight_layout()
        if is_ipython:
            display_handle = display.display(fig, display_id=True)

    # Compute smooth 50-episode moving average (smoother curve)
    means = np.zeros(len(durations_t))
    for i in range(len(durations_t)):
        start_idx = max(0, i - 49)
        means[i] = np.mean(durations_t[start_idx : i + 1].numpy())

    # Update line data directly in-place
    line_raw.set_data(episodes, durations_t.numpy())
    line_avg.set_data(episodes, means)

    # Smoothly expand x-axis if episodes exceed current limit
    if len(episodes) > ax.get_xlim()[1] - 5:
        ax.set_xlim(0, len(episodes) + 50)

    # Update title with latest episode statistics
    if show_result:
        final_avg = means[-1] if len(means) > 0 else 0
        ax.set_title(f'Result - Final 50-Ep Avg Duration: {final_avg:.1f} steps', fontsize=13, fontweight='bold')
    else:
        current_avg = means[-1] if len(means) > 0 else 0
        ax.set_title(
            f'Training DQN (CartPole-v1)... Ep {len(episodes)} | 50-Ep Avg: {current_avg:.1f} | Last: {int(durations_t[-1].item())}',
            fontsize=13,
            fontweight='bold'
        )

    # In-place DOM update via IPython display handle (completely flicker-free)
    if is_ipython:
        if display_handle is not None:
            display_handle.update(fig)
        else:
            display.clear_output(wait=True)
            display.display(fig)
    else:
        plt.pause(0.001)

    if show_result:
        plt.close(fig)
```

<div style="background: linear-gradient(90deg, rgba(56, 189, 248, 0.18) 0%, rgba(99, 102, 241, 0.10) 55%, transparent 100%); border-left: 5px solid #38bdf8; padding: 10px 18px; border-radius: 8px; margin: 24px 0 14px 0;">
  <h2 style="margin: 0; font-size: 1.35rem; font-weight: 700; letter-spacing: -0.01em;">
    6. Optimization: Loss Function & Gradient Descent
  </h2>
</div>

For each sampled mini-batch:
1. **Predicted Q-Value:** $Q(s_t, a_t; \theta)$ extracted via `.gather(1, action_batch)`.
2. **Bellman Target:**
   $$y_t = r_t + \gamma \max_{a'} Q_{\text{target}}(s_{t+1}, a') \cdot (1 - \text{done}_t)$$
3. **Huber Loss (Smooth L1):** Quadratic for small errors, linear for large errors to prevent gradient explosion.
$$
\begin{array}{c} L_\delta(y, \hat{y}) =  \begin{cases}  \frac{1}{2}(y - \hat{y})^2 & \text{for } \vert{}y - \hat{y}\vert{} \le \delta \\ \delta \cdot \left(\vert{}y - \hat{y}\vert{} - \frac{1}{2}\delta\right) & \text{for } \vert{}y - \hat{y}\vert{} > \delta \end{cases} \end{array}
$$

<p align="center">
  <img src="huber_loss.png" width="500">
</p>

```python
def optimize_model():
    """Sample a mini-batch from replay memory and perform one gradient update on policy_net."""
    if len(memory) < BATCH_SIZE:
        return None

    # Transpose the batch (see https://stackoverflow.com/a/19343/3343043 for detailed explanation).
    # This converts batch-array of Transitions to Transition of batch-arrays.
    transitions = memory.sample(BATCH_SIZE)
    batch = Transition(*zip(*transitions))

    # Mask for non-final next states
    # (a final state would've been the one after which simulation ended)
    non_final_mask = torch.tensor(
        [s is not None for s in batch.next_state],
        device=device,
        dtype=torch.bool,
    )
    non_final_next_states = [s for s in batch.next_state if s is not None]

    state_batch = torch.cat(batch.state)
    action_batch = torch.cat(batch.action)
    reward_batch = torch.cat(batch.reward)

    # 1. Compute Q(s_t, a_t) for actions actually taken
    state_action_values = policy_net(state_batch).gather(1, action_batch)

    # 2. Compute Bellman target y_t using target_net
    next_state_values = torch.zeros(BATCH_SIZE, device=device)
    if non_final_next_states:
        with torch.no_grad():
            next_state_values[non_final_mask] = (
                target_net(torch.cat(non_final_next_states)).max(dim=1).values
            )

    expected_state_action_values = reward_batch + (GAMMA * next_state_values)

    # 3. Compute Huber Loss and update weights
    loss = criterion(state_action_values, expected_state_action_values.unsqueeze(1))
    optimizer.zero_grad()
    loss.backward()
    torch.nn.utils.clip_grad_value_(policy_net.parameters(), 100)
    optimizer.step()

    return loss.item()
```

<div style="background: linear-gradient(90deg, rgba(56, 189, 248, 0.18) 0%, rgba(99, 102, 241, 0.10) 55%, transparent 100%); border-left: 5px solid #38bdf8; padding: 10px 18px; border-radius: 8px; margin: 24px 0 14px 0;">
  <h2 style="margin: 0; font-size: 1.35rem; font-weight: 700; letter-spacing: -0.01em;">
    7. Training the DQN Agent
  </h2>
</div>

At each timestep:
1. Select action via $\epsilon$-greedy policy (`select_action`).
2. Step environment in the live simulation window and save transition to `ReplayMemory`.
3. Sample a mini-batch and update `policy_net` (`optimize_model`).
4. Softly update `target_net`: $\theta_{\text{target}} \leftarrow \tau \theta_{\text{policy}} + (1-\tau)\theta_{\text{target}}$.
5. At the end of each episode, update the live reward (duration) plot via `plot_durations()`.
6. Decay $\epsilon$ after each episode.

```python
# Reset plot tracking for a fresh run
fig, ax = None, None
line_raw, line_avg = None, None
display_handle = None
episode_durations.clear()

env = gym.make("CartPole-v1", render_mode="human")  # Recreate environment for training

if torch.cuda.is_available():
    num_episodes = 200
else:
    num_episodes = 100

current_epsilon = EPS_START

for i_episode in range(num_episodes):
    # Initialize the environment and get its state
    state, info = env.reset()
    state = torch.tensor(state, dtype=torch.float32, device=device).unsqueeze(0)

    for t in count():
        action = select_action(state, current_epsilon)
        observation, reward, terminated, truncated, _ = env.step(action.item())
        reward_tensor = torch.tensor([reward], device=device, dtype=torch.float32)
        done = terminated or truncated

        if terminated:
            next_state = None
        else:
            next_state = torch.tensor(observation, dtype=torch.float32, device=device).unsqueeze(0)

        # Store the transition in memory
        memory.push(state, action, next_state, reward_tensor, terminated)

        # Move to the next state
        state = next_state

        # Perform one step of the optimization (on the policy network)
        optimize_model()

        # Soft update of the target network's weights
        with torch.no_grad():
            for target_param, policy_param in zip(target_net.parameters(), policy_net.parameters()):
                target_param.mul_(1.0 - TAU)
                target_param.add_(TAU * policy_param)

        if done:
            episode_durations.append(t + 1)
            plot_durations()
            break

    # Decay exploration rate after each episode
    current_epsilon = max(EPS_END, current_epsilon * EPS_DECAY)

print('Complete')
plot_durations(show_result=True)
```

```python
# Close environment cleanly
env.close()
print("Environment closed cleanly.")
```
